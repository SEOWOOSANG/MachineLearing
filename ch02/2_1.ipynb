```python
import sklearn
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
```

```python
import pandas as pd

iris = load_iris()

iris_data = iris.data   # 피처: 꽃받침의 너비, 길이, 꽃잎의 너비, 길이
iris_label = iris.target    # 아이리스 품종: 정답

print("iris target값:", iris_label)
print("iris target명:", iris.target_names)
```

```python
iris.data
```

```python
iris.target_names
```

```python
iris_df = pd.DataFrame(data = iris_data, columns = iris.feature_names)
iris_df["label"] = iris.target
iris_df.head()
```

```python
# 학습셋과 테스트셋으로 나누기
X_train, X_test, y_train, y_test = train_test_split(iris_data, iris_label,
                                                     test_size = 0.2, random_state = 11)
# test_size = 0.2면 train_size는 0.8
```

```python
dt_clf = DecisionTreeClassifier(random_state = 11)  # 결정트리 클래스 객체 생성
```

```python
dt_clf.fit(X_train, y_train)    # X_train, y_train 값을 dt_clf에 할당
```

```python
pred = dt_clf.predict(X_test)   
```

```python
pred
```

```python
from sklearn.metrics import accuracy_score
print("예측 정확도: ", accuracy_score(y_test, pred))
```

```python
from sklearn.model_selection import KFold
import numpy as np

kfold = KFold(n_splits = 5)     # 데이터 값을 5개로 나눔
cv_accuracy = []
```

```python
n_iter = 0
features = iris.data
label = iris.target

for train_index, test_index in kfold.split(features):               # iris.data를 kfold로 나눔
    X_train, X_test = features[train_index], features[test_index]   # 학습 데이터와 검증 데이터 값을 X_train, X_test에 할당
    y_train, y_test = label[train_index], label[test_index]         # label 값을 y_train, y_test에 할당

    dt_clf.fit(X_train, y_train)                                    # 학습 데이터와 검증 데이터 값 df_clf에 할당
    pred = dt_clf.predict(X_test)
    n_iter += 1                                                     

    accuracy = np.round(accuracy_score(y_test, pred), 4)            # accuracy 값을 소수점 4자리까지 출력
    train_size = X_train.shape[0]                                   # 학습 데이터 행 개수 확인
    test_size = X_test.shape[0]                                     # 검증 데이터 행 개수 확인
    print(n_iter)
    print("교차검증 정확도: ", accuracy)
    print("학습 데이터 크기: ", train_size)
    print("검증 데이터 크기: ", test_size)
    cv_accuracy.append(accuracy)

print("평균 정확도: ", np.mean(cv_accuracy))
```

```python
from sklearn.model_selection import StratifiedKFold

skf = StratifiedKFold(n_splits = 3) # 데이터를 3개로 나눔
n_iter = 0

for train_index, test_index in skf.split(iris_df, iris_df["label"]):    # iris_df를 학습 인덱스와 검증 인덱스로 나눔
    n_iter += 1 
    label_train = iris_df["label"].iloc[train_index]
    label_test = iris_df["label"].iloc[test_index]
    print("학습 레이블 데이터 분포: ", label_train.value_counts())  # 학습 인덱스로 label 가져옴
    print("검증 레이블 데이터 분포: ", label_test.value_counts())   # 검증 인덱스로 label 가져옴
```

```python
df_clf = DecisionTreeClassifier(random_state = 156) # 의사 결정트리 모델 생성
skfold = StratifiedKFold(n_splits = 3)  # 데이터를 3개로 나눔
n_iter = 0
cv_accuracy = []

for train_index, test_index in skfold.split(features, label):   # features와 label을 이용해 학습 인덱스와 검증 인덱스로 나눔
    X_train, X_test = features[train_index], features[test_index]   # 학습 인덱스와 검증 인덱스를 이용해 X_train, X_test를 나눔
    y_train, y_test = label[train_index], label[test_index]         # 학습 인덱스와 검증 인덱스를 ㅇ용해 y_train, y_test를 나눔
    df_clf.fit(X_train, y_train)
    pred = df_clf.predict(X_test)

    n_iter += 1
    accuracy = np.round(accuracy_score(y_test, pred), 4)        # 소수점 4자리까지 출력
    cv_accuracy.append(accuracy)

print("평균 검증 정확도: ", np.round(np.mean(cv_accuracy), 4))      # 소수점 4자리까지 출력
```

```python
from sklearn.model_selection import cross_val_score

scores = cross_val_score(dt_clf, features, label, scoring = "accuracy", cv = 3) # 데이터를 3개로 나눔
print("교차 검증별 정확도: ", np.round(scores, 4))  # 소수점 4자리까지 출력
print("평균 검증별 정확도:", np.round(np.mean(scores), 4))  # 소수점 4자리까지 출력
```

```python
from sklearn.model_selection import GridSearchCV
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target,
                                                   test_size = 0.2, random_state = 121) # test_size = 0.2면 train_size는 0.8

parameters = {"max_depth": [1, 2, 3], "min_samples_split": [2, 3]}
```

```python
dtree = DecisionTreeClassifier()

grid_dtree = GridSearchCV(dtree, param_grid = parameters, cv = 3, refit = True)
grid_dtree.fit(X_train, y_train)
scores_df = pd.DataFrame(grid_dtree.cv_results_)
scores_df[["params", "mean_test_score", "rank_test_score", "split0_test_score", "split1_test_score", "split2_test_score"]]
```

```python
grid_dtree.best_params_
```

```python
grid_dtree.best_score_
```

```python
estimator = grid_dtree.best_estimator_

pred = estimator.predict(X_test)
print(accuracy_score(y_test, pred))
```

```python
from sklearn.datasets import load_iris

iris = load_iris()
iris_data = iris.data   # 데이터 저장
iris_df = pd.DataFrame(data = iris_data, columns = iris.feature_names)  # 데이터 프레임 생성

print(iris_df.mean())
print(iris_df.var())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler() # 스케일러 생성
scaler.fit(iris_df) # 데이터 평균과 표준편차 학습
iris_scaled = scaler.transform(iris_df) # 데이터 표준화

iris_df_scaled = pd.DataFrame(data = iris_scaled, columns = iris.feature_names) # 데이터 프레임 생성
print(iris_df_scaled.mean())
print(iris_df_scaled.var())
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler() # 스케일러 생성
scaler.fit(iris_df) # 데이터, 최솟값, 최댓값 학습
iris_scaled = scaler.transform(iris_df) # 데이터를 0~1 범위로 변환

iris_df_scaled = pd.DataFrame(data = iris_scaled, columns = iris.feature_names) # 데이터 프레임 생성
print(iris_df_scaled.min())
print(iris_df_scaled.max())
```
